```python
def chatbot():
    print("Hello!")

    while True:
        n = input("You: ").lower()
        if n == "quit":
            print("Chatbot: Goodbye!")
            break

        response_text = "I'm sorry, I don't understand that question."
        if n == "hello":
            response_text = "Hi"
        elif n == "how are you":
            response_text = " I'm ready to assist you!"
        elif n == "what is your name":
            response_text = "I am a simple chatbot."
        elif n == "bye":
            response_text = "Goodbye! "
        elif n == "thanks":
            response_text = "You're welcome!"
        print(f"Chatbot: {response_text}")

chatbot()
```

```python
g = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}
v = []
q = ['A']
while q:
    n = q.pop(0)
    if n not in v:
        print(n, end=" ")
        v.append(n)
        q += g[n]
```

```python
g = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

v = []

def dfs(n):
    if n not in v:
        print(n, end=" ")
        v.append(n)
        for x in g[n]:
            dfs(x)

dfs('A')
```
