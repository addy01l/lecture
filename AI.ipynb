```python
def chatbot():
    print("Hello!")

    while True:
        n = input("You: ").lower()
        if n == "quit":
            print("Chatbot: Goodbye!")
            break

        response_text = "I'm sorry, I don't understand that question."
        if n == "hello":
            response_text = "Hi"
        elif n == "how are you":
            response_text = " I'm ready to assist you!"
        elif n == "what is your name":
            response_text = "I am a simple chatbot."
        elif n == "bye":
            response_text = "Goodbye! "
        elif n == "thanks":
            response_text = "You're welcome!"
        print(f"Chatbot: {response_text}")

chatbot()
```

```python
g = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

v = []

def dfs(n):
    if n not in v:
        print(n, end=" ")
        v.append(n)
        for x in g[n]:
            dfs(x)

dfs('A')
```

```python
def h(a):
    goal = [[1,2,3],[4,5,6],[7,8,0]]
    return sum(a[i][j] != 0 and a[i][j] != goal[i][j]
               for i in range(3) for j in range(3))

def hill(a):
    while h(a):
        x, y = next((i,j) for i in range(3) for j in range(3) if a[i][j] == 0)
        best = None

        for dx, dy in [(1,0),(-1,0),(0,1),(0,-1)]:
            p, q = x+dx, y+dy
            if 0 <= p < 3 and 0 <= q < 3:
                b = [r[:] for r in a]
                b[x][y], b[p][q] = b[p][q], b[x][y]
                if best is None or h(b) < h(best):
                    best = b

        if h(best) >= h(a):
            print("Stuck")
            return
        a = best
        for r in a:
            print(r)
        print()

hill([[1,2,3],[0,4,6],[7,5,8]])
```
