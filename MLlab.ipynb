```python
data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same','same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'same','Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change','change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change','change' ,'Yes']
]

n = len(data[0]) - 1

S = ['0'] * n
G = [['?'] * n]

for row in data:
    x = row[:-1]
    y = row[-1]

    if y == 'Yes':
        for i in range(n):
            if S[i] == '0':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'

        G = [g for g in G if all(g[i] == '?' or g[i] == S[i] for i in range(n))]

    else:
        new_g = []

        for g in G:
            for i in range(n):
                if g[i] == '?' and S[i] != '?' and S[i] != x[i]:
                    h = g.copy()
                    h[i] = S[i]
                    new_g.append(h)

        G = new_g

print("Specific:", S)
print("General:", G)
```

```python
import csv
from collections import defaultdict

def load_data(filepath):
    with open(filepath, 'r') as file:
        reader = csv.reader(file)
        data = list(reader)

    return data[0], data[1:]


def train_naive_bayes(data):
    class_count = defaultdict(int)
    feature_count = defaultdict(lambda: defaultdict(int))
    feature_values = defaultdict(set)

    for row in data:
        features = row[:-1]
        label = row[-1]

        class_count[label] += 1

        for i, value in enumerate(features):
            feature_count[label][(i, value)] += 1
            feature_values[i].add(value)

    return class_count, feature_count, feature_values


def predict(row, class_count, feature_count, feature_values, total):
    best_class = None
    best_probability = -1

    for label in class_count:
        probability = class_count[label] / total

        for i, value in enumerate(row):
            count = feature_count[label].get((i, value), 0)

            probability *= (
                count + 1
            ) / (class_count[label] + len(feature_values[i]))

        if probability > best_probability:
            best_probability = probability
            best_class = label

    return best_class


# Load training data
header, train_data = load_data("/content/data11.csv")

# Train model
class_count, feature_count, feature_values = train_naive_bayes(train_data)

# Test data
test_data = [
    ["Sunny", "Cool", "High", "True", "No"],
    ["Overcast", "Mild", "High", "False", "Yes"],
    ["Rain", "Cool", "Normal", "False", "Yes"],
    ["Sunny", "Mild", "Normal", "True", "Yes"],
    ["Rain", "Mild", "High", "True", "No"],
]

correct = 0

print("Predictions")
print("-" * 40)

for row in test_data:
    features = row[:-1]
    actual = row[-1]

    predicted = predict(
        features, class_count, feature_count, feature_values, len(train_data)
    )

    print("Actual:", actual, "| Predicted:", predicted)

    if actual == predicted:
        correct += 1

accuracy = (correct / len(test_data)) * 100

print("-" * 40)
print("Correct Predictions:", correct)
print("Total Test Cases:", len(test_data))
print("Accuracy:", accuracy, "%")
```
