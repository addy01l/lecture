```python
import numpy as np
ar = np.array([1, 2, 3, 4, 5])
ar2 = np.array([6, 7, 8, 9, 10])
print(ar + ar2)
print( ar - ar2)
print(ar *ar2)
print(ar/ar2)
print(np.add(ar ,ar2))
print(np.subtract(ar,ar2))
print(np.multiply(ar,ar2))
print(np.divide(ar,ar2))
```
