```python
sentence = "the students are reading books"
word = sentence.replace(".","").split()
print("Words :",word)
word_knowledge = {
    "students" : "noun",
    "reading" : "verb",
    "books" : "noun"
}
for w in word:
  key = w.lower()
  if key in word_knowledge:
    print(w , "->",word_knowledge [key])
```

```python
text = "My order has not arrived yet."
words = text.lower().replace(".", "").split()
keywords = ["order", "delivery", "arrived",
"refund"]
for word in words:
  if word in keywords:
   print("Important word:", word)
```
